```python
import pandas as pd
from sqlalchemy import create_engine

CSV_FILE = "../data/yellow_tripdata_2024-01.csv"
TABLE_NAME = "yellow_taxi_data"

engine = create_engine(
    "postgresql://postgres:changeme1234@localhost:5432/postgres"
)

total_loaded = 0

for chunk_number, chunk in enumerate(
    pd.read_csv(
        CSV_FILE,
        dtype=str,
        chunksize=10_000
    )
):
    chunk.to_sql(
        name=TABLE_NAME,
        con=engine,
        if_exists="replace" if chunk_number == 0 else "append",
        index=False
    )

    total_loaded += len(chunk)

    print(
        f"Bloque {chunk_number + 1}: "
        f"{len(chunk)} filas | "
        f"total acumulado: {total_loaded}"
    )
```
