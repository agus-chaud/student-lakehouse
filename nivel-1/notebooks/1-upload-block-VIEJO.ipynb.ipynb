```python
import pandas as pd
from sqlalchemy import create_engine

CSV_FILE = "../data/yellow_tripdata_2024-01.csv"
TABLE_NAME = "yellow_taxi_data"
```

```python
engine = create_engine(
    "postgresql://postgres:changeme1234@localhost:5432/postgres"
)
```

```python

for chunk_number, chunk in enumerate(
    pd.read_csv(
        CSV_FILE,
        dtype=str,
        chunksize=10_000
    )
):
    chunk.to_sql(
        name=TABLE_NAME,
        con=engine,
        if_exists="replace" if chunk_number == 0 else "append",
        index=False
    )

    print(f"Bloque {chunk_number + 1} cargado: {len(chunk)} filas")
```
